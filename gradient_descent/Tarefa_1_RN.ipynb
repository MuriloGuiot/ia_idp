# **Construa o Modelo**

---
Crie o seu modelo

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report

# --- Preparação dos Dados ---
path = 'pima_indians_diabetes.csv'
df = pd.read_csv(path)
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# --- Construção e Treinamento do Modelo ---
model = MLPClassifier(
    hidden_layer_sizes=(12, 8),
    activation='relu',
    solver='adam',
    max_iter=1000,
    learning_rate_init=0.001,
    random_state=42
)

print("Iniciando o treinamento...")
model.fit(X_train, Y_train)
print("Treinamento concluído!")

# --- Avaliação e Exibição dos Resultados ---
print("\n--- Avaliação do Modelo ---")

# Fazendo previsões nos dados de teste
Y_pred = model.predict(X_test)

# Calculando a precisão do modelo
accuracy = accuracy_score(Y_test, Y_pred)
print(f"Acurácia do modelo nos dados de teste: {accuracy:.4f}")

# Exibindo um relatório de classificação detalhado
report = classification_report(Y_test, Y_pred)
print("\nRelatório de Classificação:\n", report)

# Exibindo as previsões e os valores reais para as primeiras 10 amostras
print("\nPrimeiras 10 previsões vs. valores reais:")
for i in range(10):
    print(f"Previsão: {Y_pred[i]}, Real: {Y_test.iloc[i]}")
```

---
Compile o modelo escolhendo a função de perda e o otimizador.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Preparação dos Dados
path = 'pima_indians_diabetes.csv'
df = pd.read_csv(path)
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# Função de Perda:
# Para a classificação, o MLPClassifier usa a "Cross-Entropy"
# (Entropia Cruzada) por padrão. O Scikit-learn gerencia isso
# internamente, pois é a função de perda mais adequada para este
# tipo de problema. Ela mede quão erradas estão as previsões
# do modelo em relação aos valores reais.

# Otimizador (Gradient Descent):
# O parâmetro 'solver' é onde você escolhe o otimizador.
# 'adam' é uma versão otimizada do Gradiente Descendente.
# Ele ajusta a taxa de aprendizado de forma adaptativa,
# tornando o treinamento mais rápido e robusto.

model = MLPClassifier(
    hidden_layer_sizes=(12, 8),
    activation='relu',
    solver='adam',  # Este é o OTIMIZADOR
    max_iter=1000,
    learning_rate_init=0.001,
    random_state=42
)

# 3. Treinar o Modelo
print("Iniciando o treinamento do modelo...")
model.fit(X_train, Y_train)
print("Treinamento concluído!")

# 4. Avaliar o Modelo
print("\n--- Avaliação do Modelo ---")
Y_pred = model.predict(X_test)
accuracy = accuracy_score(Y_test, Y_pred)
print(f"Acurácia do modelo nos dados de teste: {accuracy:.4f}")
report = classification_report(Y_test, Y_pred)
print("\nRelatório de Classificação:\n", report)
```

Avalie a precisão usando o comando `model.evaluate()`

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier

# 1. Preparação dos Dados
path = 'pima_indians_diabetes.csv'
df = pd.read_csv(path)
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# 2. Construção e Treinamento do Modelo
model = MLPClassifier(
    hidden_layer_sizes=(12, 8),
    activation='relu',
    solver='adam',
    max_iter=1000,
    learning_rate_init=0.001,
    random_state=42
)

print("Iniciando o treinamento do modelo...")
model.fit(X_train, Y_train)
print("Treinamento concluído!")

# 3. Avaliar a precisão com model.score()
accuracy = model.score(X_test, Y_test)

print("\n--- Avaliação do Modelo ---")
print(f"A precisão do modelo nos dados de teste é: {accuracy:.4f}")
```

Crie uma pessoa imaginária com um vetor, por exemplo, na forma `np.array([[2,148,72,35,0,33.6,0.627,25]])` e use o comando `model.predict()` para verificar se essa pessoa é classificada com ou sem diabetes.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier

path = 'pima_indians_diabetes.csv'
df = pd.read_csv(path)
X = df.iloc[:, :-1]
Y = df.iloc[:, -1]
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = MLPClassifier(
    hidden_layer_sizes=(12, 8),
    activation='relu',
    solver='adam',
    max_iter=1000,
    learning_rate_init=0.001,
    random_state=42
)

model.fit(X_train, Y_train)

nova_pessoa = np.array([[2, 148, 72, 35, 0, 33.6, 0.627, 25]])

previsao = model.predict(nova_pessoa)

print(f"Dados de entrada da nova pessoa: {nova_pessoa[0]}")
print(f"Previsão do modelo: {previsao[0]}")

if previsao[0] == 1:
    print("O modelo classifica esta pessoa como tendo diabetes.")
else:
    print("O modelo classifica esta pessoa como não tendo diabetes.")
```
