```python
import numpy as np
import matplotlib.pyplot as plt

# A função p minimizar
def f(x):
    return x**2

# A derivada da função
def df(x):
    return 2 * x

taxa_aprendizagem = 0.1  # O tamanho de cada "passo" dado 
iteracoes = 25           # Quantos passos vamos dar
ponto_inicial = 10       # Onde começar na curva

historico_pontos = [ponto_inicial]
x_atual = ponto_inicial

for i in range(iteracoes):
    gradiente = df(x_atual)  # Calcula a inclinação no ponto atual
    # Move na direção oposta ao gradiente (para descer)
    x_atual = x_atual - taxa_aprendizagem * gradiente
    historico_pontos.append(x_atual)
    print(f"Iteração {i+1}: x = {x_atual:.4f}, f(x) = {f(x_atual):.4f}")

print(f"\nO ponto mínimo encontrado é aproximadamente: {x_atual:.4f}")

x_vals = np.linspace(-11, 11, 400)
y_vals = f(x_vals)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label='Função f(x) = x²')
plt.plot(historico_pontos, f(np.array(historico_pontos)), 'ro-', label='Passos do Gradiente Descendente')

plt.title('Visualização do Gradiente Descendente')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```
