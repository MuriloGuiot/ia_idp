```python
# Importar Librerias
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error

#Importar Dataset
ruta = 'housing_prices_California.xlsx'
database = pd.read_excel(ruta)

# Pré-processamento de dados (seleção de recursos relevantes e alvo)
X = database[['median_income', 'total_rooms', 'housing_median_age']]
y = database['median_house_value']
# Dividido em conjuntos de treinamento e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, shuffle=False, test_size=0.3)
```

```python
# Defina a grade de parâmetros para pesquisar
param_grid = {
'C': [0.1, 1, 10, 100],
'gamma': ['scale', 'auto'],
'kernel': ['linear', 'rbf', 'poly']
}

 # Executar pesquisa em grade com validação cruzada
grid_search = GridSearchCV(SVR(), param_grid, cv=5, scoring='neg_mean_squared_error')
grid_search.fit(X_train, y_train)

 # Melhor combinação de parâmetros
print(f'Best hyperparameters: {grid_search.best_params_}')

 # Prever no conjunto de teste usando o melhor modelo encontrado
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)

 # Avaliar o modelo
mse = mean_squared_error(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
```
